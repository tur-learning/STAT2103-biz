# **📌 Chapter 7: Inferences Based on a Single Sample - Hypothesis Testing**


---
## **🔹 7.1 The Elements of a Test of Hypothesis**


### **1️⃣ What is Hypothesis Testing?**
Hypothesis testing is a method used to make **decisions** about a population **based on sample data**. Instead of estimating a parameter (like in confidence intervals), we **test a claim (hypothesis) about the parameter**.

📌 **Example Scenario:**  
A city requires that sewer pipes must have an average **breaking strength** greater than **2,400 pounds per foot**. A manufacturer must demonstrate that their pipes meet this specification.

---
### **2️⃣ Null vs. Alternative Hypothesis**
- **Null Hypothesis** ($ H_0 $)  
  The default assumption, **status quo**, or claim we are testing.
  - Example: The pipes **do not meet** the required strength.  
    \begin{align}
    H_0: \mu \leq 2400
    \end{align}
- **Alternative (research) Hypothesis** ($ H_a $)  
  What we accept if the **null hypothesis is rejected**.
  - Example: The pipes **meet specifications** (stronger than 2400).  
    \begin{align}
    H_a: \mu > 2400
    \end{align}

📌 **Key Takeaway:**  
- If **data strongly contradict $ H_0 $**, we **reject** it and accept $ H_a $.
- If **data does not contradict $ H_0 $**, we **fail to reject** it (but we don’t "accept" it).

---
### **3️⃣ Test Statistic**
A **test statistic** is a value computed from sample data that helps decide whether to reject $ H_0 $.

📌 **For means (large samples, known $ \sigma $):**
\begin{align}
z = \frac{\bar{x} - \mu_0}{\sigma / \sqrt{n}}
\end{align}
Where:
- $ \bar{x} $ = Sample mean
- $ \mu_0 $ = Hypothesized population mean
- $ \sigma $ = Population standard deviation
- $ n $ = Sample size

---
### **4️⃣ Rejection Region & Significance Level ($ \alpha $)**
- We compare the test statistic ($ z $-score) to a **critical value** ($ z_{\alpha} $) from the normal table.
- The **level of significance** ($ \alpha $) is the probability of **rejecting $ H_0 $ when it's actually true** (**Type I Error**).
- Common choices:
  - **$ \alpha = 0.05 $** → 5% chance of Type I error
  - **$ \alpha = 0.01 $** → 1% chance (stricter)
  - **$ \alpha = 0.10 $** → 10% chance (more lenient)

📌 **Example:**
If $ \alpha = 0.05 $ and we have a **one-tailed test**, the **critical value** is:
- $ z_{0.05} = 1.645 $ for a **right-tailed test**
- $ z_{0.05} = -1.645 $ for a **left-tailed test**

---
### **5️⃣ Type I and Type II Errors**
| Decision | True $ H_0 $ | True $ H_a $ |
|------------|------------|------------|
| Reject $ H_0 $ | **Type I Error** (with probability $ \alpha $) | ✅ Correct Decision |
| Fail to Reject $ H_0 $ | ✅ Correct Decision | **Type II Error** (with probability $ \beta $) |

📌 **Interpretation:**
- **Type I Error (with probability $ \alpha $)**: Rejecting a true $ H_0 $ (False Alarm).
- **Type II Error (with probability $ \beta $)**: Failing to reject a false $ H_0 $ (Missed Opportunity).

---

## **🔹 Example 7.1: Testing Sewer Pipe Strength**
A city requires sewer pipes to have **$ \mu > 2400 $ pounds per foot**.

📌 **Given Data:**
- **Sample size:** $ n = 50 $
- **Sample mean:** $ \bar{x} = 2430 $
- **Sample standard deviation:** $ s = 200 $
- **Significance level:** $ \alpha = 0.05 $
- **Null hypothesis:** $ H_0: \mu \leq 2400 $
- **Alternative hypothesis:** $ H_a: \mu > 2400 $ (Right-tailed test)

📌 **Step 1: Compute the test statistic**
\begin{align}
z = \frac{\bar{x} - \mu_0}{s / \sqrt{n}}
\end{align}
\begin{align}
= \frac{2430 - 2400}{200 / \sqrt{50}}
\end{align}
\begin{align}
= \frac{30}{28.28} = 1.06
\end{align}

📌 **Step 2: Determine critical value**  
- From **Z-table**, for **$ \alpha = 0.05 $ (right-tailed test)**:  
  \begin{align}
  z_{0.05} = 1.645
  \end{align}

📌 **Step 3: Compare test statistic vs. critical value**  
- $ z = 1.06 $ **<** $ z_{0.05} = 1.645 $  
- **Conclusion:** The test statistic **is not in the rejection region** → **Fail to reject $ H_0 $**.

✅ **Interpretation:**  
There is **insufficient evidence** to conclude that the pipes meet the **2400-pound specification**.

---
## **🔹 Steps in Hypothesis Testing**
1️⃣ **State the Hypotheses**  
   - $ H_0: \mu = \mu_0 $ (Null Hypothesis)  
   - $ H_a: \mu \neq \mu_0 $ (Two-Tailed) or $ \mu > \mu_0 $ (Right-Tailed)  

2️⃣ **Select Significance Level ($ \alpha $)**  
   - Common choices: **$ 0.05, 0.01, 0.10 $**

3️⃣ **Compute the Test Statistic**  
   - $ z = \frac{\bar{x} - \mu_0}{\sigma / \sqrt{n}} $ (if $ \sigma $ known)
   - $ t = \frac{\bar{x} - \mu_0}{s / \sqrt{n}} $ (if $ \sigma $ unknown, small $ n $)

4️⃣ **Determine Critical Value(s)**
   - **Z-table** (large sample, known $ \sigma $)
   - **t-table** (small sample, unknown $ \sigma $)

5️⃣ **Make a Decision**
   - If $ |z| > z_{\alpha/2} $ → **Reject $ H_0 $**
   - If $ |z| \leq z_{\alpha/2} $ → **Fail to reject $ H_0 $**

---

## **🔹 Python Implementation**
### **✅ Automate Hypothesis Testing**

```python
import scipy.stats as stats
import numpy as np

# Given data
x_bar = 2430  # Sample mean
mu_0 = 2400  # Hypothesized mean
s = 200  # Sample standard deviation
n = 50  # Sample size
alpha = 0.05  # Significance level

# Compute test statistic
z_score = (x_bar - mu_0) / (s / np.sqrt(n))

# Find critical value
z_critical = stats.norm.ppf(1 - alpha)  # One-tailed test

# Decision
if z_score > z_critical:
    print(f"Reject H0 (z = {z_score:.2f} > {z_critical:.2f}) → Evidence supports alternative hypothesis")
else:
    print(f"Fail to reject H0 (z = {z_score:.2f} <= {z_critical:.2f}) → Insufficient evidence")
```

---
## **🔹 Summary**
📌 **Hypothesis testing** helps decide whether sample data provide strong enough evidence against $ H_0 $.  
📌 **Type I Error (with probability $ \alpha $)**: Rejecting a true $ H_0 $.  
📌 **Type II Error (with probability $ \beta $)**: Failing to reject a false $ H_0 $.  
📌 **Decision Rule**: Compare **test statistic** vs. **critical value**.  
📌 **$ \alpha = 0.05 $** is commonly used for **business and engineering**.

# **📌 7.2 Formulating Hypotheses and Setting Up the Rejection Region**


---

## **🔹 1. Understanding Hypothesis Formulation**
In hypothesis testing, we define two competing hypotheses:

1️⃣ **Null Hypothesis ($ H_0 $)**  
   - Represents the status quo, or a claim to be tested.  
   - Always stated with **equality**: $ =, \leq, \geq $.  
   - We assume $ H_0 $ is **true** until evidence suggests otherwise.  

2️⃣ **Alternative Hypothesis ($ H_a $)**  
   - Represents what we **want to prove** (the research hypothesis).  
   - Always stated with **strict inequality**: $ \neq, >, < $.  

## **🔹 2. One-Tailed vs. Two-Tailed Tests**
Depending on the research question, we conduct a **one-tailed or two-tailed** test.

| **Test Type** | **Alternative Hypothesis ($ H_a $)** | **Rejection Region** | **Example Keywords** |
|--------------|--------------------------------|----------------|--------------------|
| **Right-tailed (upper-tailed)** | $ H_a: \mu > \mu_0 $ | Large values of test statistic | "greater than," "above" |
| **Left-tailed (lower-tailed)** | $ H_a: \mu < \mu_0 $ | Small values of test statistic | "less than," "below" |
| **Two-tailed** | $ H_a: \mu \neq \mu_0 $ | Large or small values | "not equal to," "differs from" |

📌 **Example**:  
- Testing if a machine fills cereal boxes with **exactly 12 ounces**.  
  - If deviations **above or below** 12 oz are problematic: **Two-tailed test**.  
  - If only **underfilling** is an issue: **Left-tailed test**.

---

## **🔹 3. Examples of Hypothesis Formulation**


### **📌 Example 1: Testing a Lathe Machine (Two-Tailed)**
A **metal lathe** produces bearings, and quality control checks if the **mean diameter is 0.5 inches**. If the mean diameter is too **large or small**, adjustments are needed.

📌 **Formulating Hypotheses:**
- **$ H_0: \mu = 0.5 $** (Lathe is working correctly)  
- **$ H_a: \mu \neq 0.5 $** (Lathe needs adjustment)

✅ **Two-tailed test:** We check if the mean is **too high or too low**.

---

### **📌 Example 2: Cigarette Warnings (One-Tailed)**
The **Federal Trade Commission (FTC)** claims **80%** of cigarette consumers fail to see health warnings. A tobacco marketer believes **this percentage is lower**.

📌 **Formulating Hypotheses:**
- **$ H_0: p = 0.80 $** (FTC's claim is correct)
- **$ H_a: p < 0.80 $** (Fewer consumers fail to see the warning)

✅ **One-tailed test (left-tailed):** The marketer **wants to prove** the true proportion is **less than** 80%.

---

### **📌 Example 3: Testing Cereal Box Filling (Two-Tailed)**
A **cereal company** wants to check if its **filling machine** is dispensing **exactly 12 oz** of cereal.

📌 **Formulating Hypotheses:**
- **$ H_0: \mu = 12 $** (Machine is properly filling)  
- **$ H_a: \mu \neq 12 $** (Machine is over/under-filling)  

✅ **Two-tailed test:**  
We reject $ H_0 $ if the mean fill is **too high or too low**.

---
## **🔹 4. Setting Up the Rejection Region**


### **📌 What is the Rejection Region?**
The rejection region is the range of values where we **reject $ H_0 $** in favor of $ H_a $. It is determined by:
- **The significance level** ($ \alpha $, probability of Type I error)
- **The type of test** (one-tailed or two-tailed)

### **📌 How to Find the Rejection Region**
1️⃣ **One-tailed test (Right-Tailed):**  
   - Reject $ H_0 $ if $ z > z_{\alpha} $.  
   - Example: If $ \alpha = 0.05 $, then **$ z_{\alpha} = 1.645 $** (from Z-table).

2️⃣ **One-tailed test (Left-Tailed):**  
   - Reject $ H_0 $ if $ z < -z_{\alpha} $.  
   - Example: If $ \alpha = 0.05 $, then **$ z_{\alpha} = -1.645 $**.

3️⃣ **Two-tailed test:**  
   - Reject $ H_0 $ if $ |z| > z_{\alpha/2} $.  
   - Example: If $ \alpha = 0.05 $, then **$ z_{\alpha/2} = \pm 1.960 $**.

### **📌 Common Critical Values (Z-table)**
| **$ \alpha $ Level** | **One-tailed $ z_{\alpha} $** | **Two-tailed $ z_{\alpha/2} $** |
|------------|-----------------|-----------------|
| 0.10 | 1.282 | 1.645 |
| 0.05 | 1.645 | 1.960 |
| 0.01 | 2.326 | 2.576 |

---
## **🔹 5. Example Calculation (Cereal Box Test)**
A cereal manufacturer wants to test if the **mean weight** of its cereal boxes **differs from 12 oz**.

### 📌 **Given Data:**
- Sample size: $ n = 100 $
- Sample mean: $ \bar{x} = 11.85 $
- Population standard deviation: $ \sigma = 0.5 $
- Significance level: $ \alpha = 0.01 $
- Hypotheses:  
  - $ H_0: \mu = 12 $  
  - $ H_a: \mu \neq 12 $ (Two-tailed)

### 📌 **Step 1: Compute the test statistic**
\begin{align}
z = \frac{\bar{x} - \mu_0}{\sigma / \sqrt{n}} = \frac{11.85 - 12}{0.5 / \sqrt{100}} = \frac{-0.15}{0.05} = -3.00
\end{align}

### 📌 **Step 2: Find the rejection region**  
For **$ \alpha = 0.01 $** (two-tailed test):
\begin{align}
z_{\alpha/2} = \pm 2.576
\end{align}

### 📌 **Step 3: Compare Test Statistic to Critical Value**  
- Since **$ z = -3.00 $** is **less than** $ -2.576 $, we **reject $ H_0 $**.

✅ **Conclusion:** The mean fill weight is significantly different from **12 oz**, and the manufacturer should adjust the machine.

---
## **🔹 6. Python Implementation**

```python
import scipy.stats as stats
import numpy as np

# Given data
x_bar = 11.85  # Sample mean
mu_0 = 12  # Hypothesized mean
sigma = 0.5  # Population standard deviation
n = 100  # Sample size
alpha = 0.01  # Significance level

# Compute test statistic
z_score = (x_bar - mu_0) / (sigma / np.sqrt(n))

# Find critical value for two-tailed test
z_critical = stats.norm.ppf(1 - alpha / 2)  # Two-tailed test

# Decision
if abs(z_score) > z_critical:
    print(f"Reject H0 (z = {z_score:.2f} > {z_critical:.2f}) → Machine needs adjustment")
else:
    print(f"Fail to reject H0 (z = {z_score:.2f} <= {z_critical:.2f}) → No adjustment needed")
```

## **🔹 Summary**
📌 **Formulating Hypotheses**:
- **Null ($ H_0 $)**: Status quo, equality sign.
- **Alternative ($ H_a $)**: What we want to prove.

📌 **Choosing the Right Test**:
- **Right-tailed**: $ H_a: \mu > \mu_0 $.
- **Left-tailed**: $ H_a: \mu < \mu_0 $.
- **Two-tailed**: $ H_a: \mu \neq \mu_0 $.

📌 **Rejection Region**:
- **One-tailed**: $ z > z_{\alpha} $ or $ z < -z_{\alpha} $.
- **Two-tailed**: $ |z| > z_{\alpha/2} $.

🚀 **Now you're ready to set up hypothesis tests like a pro!** 🎯

# **📌 7.3 Observed Significance Levels: p-Values**


---

## **🔹 1. What is a p-Value?**
The **p-value** is a probability that quantifies the strength of the evidence against the **null hypothesis** ($ H_0 $). It answers the question:

> *"If $ H_0 $ is true, what is the probability of obtaining a test statistic at least as extreme as the observed one?"*

📌 **Definition:**
- The **p-value** is the **probability** of obtaining a result at least as extreme as the observed test statistic, **assuming $ H_0 $ is true**.
- A **small** p-value suggests strong evidence **against $ H_0 $**.
- A **large** p-value suggests that the observed result could easily occur by random chance if $ H_0 $ is true.

---

## **🔹 2. How to Interpret the p-Value**

### **📌 Decision Rule**


1️⃣ **Choose a significance level ($ \alpha $):**  
   - Common values: **$ \alpha = 0.05 $ (5%)** or **$ \alpha = 0.01 $ (1%)**.

2️⃣ **Compare p-value to $ \alpha $:**  
   - If **$ p \leq \alpha $** → *Reject $ H_0 $* → Significant evidence for $ H_a $.  
   - If **$ p > \alpha $** → *Fail to reject $ H_0 $* → Insufficient evidence for $ H_a $.

📌 **Interpretation Examples:**
- **p-value = 0.002, $ \alpha = 0.05 $** → Reject $ H_0 $. Strong evidence against the null hypothesis.
- **p-value = 0.08, $ \alpha = 0.05 $** → Fail to reject $ H_0 $. Not enough evidence to support $ H_a $.

---

## **🔹 3. Finding the p-Value**
The p-value depends on the **type of hypothesis test**:

### **📌 One-Tailed Tests**
- **Right-tailed ($ H_a: \mu > \mu_0 $)**:  
  \begin{align}
  p = P(Z > z_{\text{obs}})
  \end{align}
- **Left-tailed ($ H_a: \mu < \mu_0 $)**:  
  \begin{align}
  p = P(Z < z_{\text{obs}})
  \end{align}

📌 **Example:**  
A right-tailed test where $ z_{\text{obs}} = 2.12 $:
- Look up **$ P(Z < 2.12) $** in the **Z-table** → **0.9830**.
- Compute **$ p $**:
  \begin{align}
  p = 1 - 0.9830 = 0.017
  \end{align}
- Since **$ p = 0.017 < 0.05 $**, we **reject $ H_0 $**.

---

### **📌 Two-Tailed Test**
- **$ H_a: \mu \neq \mu_0 $**
- Compute the p-value as:
  \begin{align}
  p = 2 \times P(Z > |z_{\text{obs}}|)
  \end{align}

📌 **Example:**  
A **two-tailed test** where $ z_{\text{obs}} = -2.5 $:
- Look up **$ P(Z < 2.5) = 0.9938 $**.
- Compute **$ p $**:
  \begin{align}
  p = 2 \times (1 - 0.9938) = 2 \times 0.0062 = 0.0124
  \end{align}
- If **$ p = 0.0124 < 0.05 $**, reject $ H_0 $.

---

## **🔹 4. Example: Comparing Rejection Region vs. p-Value**
Consider testing **$ H_0: \mu = 50 $ vs. $ H_a: \mu > 50 $** with $ \alpha = 0.05 $.

📌 **Given:**
- **$ z_{\text{obs}} = 1.70 $**

### **📌 Step 1: Compute p-Value**
- From the Z-table: **$ P(Z < 1.70) = 0.9554 $**
- Compute **p-value**:
  \begin{align}
  p = 1 - 0.9554 = 0.0446
  \end{align}

### **📌 Step 2: Compare to $ \alpha $**
- **$ p = 0.0446 < 0.05 $**  
- **Decision:** *Reject $ H_0 $, significant evidence for $ H_a $*.

### **📌 Step 3: Compare to Rejection Region**
- For **$ \alpha = 0.05 $** (right-tailed test), critical value:  
  \begin{align}
  z_{\alpha} = 1.645
  \end{align}
- Since **$ 1.70 > 1.645 $**, the test statistic is **in the rejection region**.
- **Decision:** *Reject $ H_0 $*.

✅ **Conclusion:** Both **p-value and rejection region** lead to the **same conclusion**.


---

## **🔹 5. Python Implementation**
Let's compute **p-values for one-tailed and two-tailed tests** using Python.

```python
import scipy.stats as stats

# Given test statistic
z_obs = 2.12  # Change this value for different test cases

# One-tailed test (right-tailed)
p_value_one_tailed = 1 - stats.norm.cdf(z_obs)

# Two-tailed test
p_value_two_tailed = 2 * (1 - stats.norm.cdf(abs(z_obs)))

print(f"One-Tailed p-value: {p_value_one_tailed:.4f}")
print(f"Two-Tailed p-value: {p_value_two_tailed:.4f}")

# Decision based on alpha = 0.05
alpha = 0.05
if p_value_one_tailed < alpha:
    print("Reject H0 (One-tailed)")
else:
    print("Fail to Reject H0 (One-tailed)")

if p_value_two_tailed < alpha:
    print("Reject H0 (Two-tailed)")
else:
    print("Fail to Reject H0 (Two-tailed)")
```

## **🔹 6. Key Takeaways**


### **📌 Why Use p-Values?**
1️⃣ **Gives more information than just "reject" or "fail to reject"**.  
2️⃣ **Lets readers choose their own $ \alpha $-level**.  
3️⃣ **More precise measure of evidence against $ H_0 $**.

### **📌 p-Values vs. Rejection Region**
- **Both methods lead to the same conclusion**.
- p-values are preferred in research because they allow more **flexibility**.

---

## **🔹 7. Ethics in Hypothesis Testing**
🚨 **Avoid “p-hacking”**  
Unethical researchers may **run multiple tests and pick only the significant ones**. **Choosing $ \alpha $ after computing p-values** is **unethical**. Always **set $ \alpha $ before running the test**.

| **3️⃣ Compare p-value to $ \alpha $** | Reject $ H_0 $ if $ p \leq \alpha $ |

# **📌 7.4 Test of Hypothesis About a Population Mean: Normal (z) Statistic**


---

## **🔹 1. Overview**
In this section, we test hypotheses about a **population mean $ \mu $** using the **normal (z) statistic** when:
1. **Sample size is large** ($ n \geq 30 $) → The **Central Limit Theorem (CLT)** ensures that the sampling distribution of $ \bar{x} $ is **approximately normal**, regardless of the population distribution.
2. **Population standard deviation $ \sigma $ is known**, or we approximate it with the sample standard deviation $ s $ when $ n $ is large.

📌 **Test Statistic:**
\begin{align}
z = \frac{\bar{x} - \mu_0}{s / \sqrt{n}}
\end{align}
- $ \bar{x} $ = Sample mean  
- $ \mu_0 $ = Hypothesized population mean under $ H_0 $  
- $ s $ = Sample standard deviation (used if $ \sigma $ is unknown)  
- $ n $ = Sample size  

---

## **🔹 2. Steps for Hypothesis Testing**

1️⃣ **State the hypotheses**:  
   - **Null Hypothesis** ($ H_0 $): The population mean equals a specified value.  
   - **Alternative Hypothesis** ($ H_a $): The population mean is **different, greater, or less** than the specified value.

2️⃣ **Set significance level $ \alpha $** (commonly 0.05 or 0.01).

3️⃣ **Compute test statistic $ z $**:
   \begin{align}
   z = \frac{\bar{x} - \mu_0}{s / \sqrt{n}}
   \end{align}

4️⃣ **Find the p-value**:
   - **One-tailed test**: $ p = P(Z > z) $ or $ p = P(Z < z) $.
   - **Two-tailed test**: $ p = 2 \times P(Z > |z|) $.

5️⃣ **Compare p-value to $ \alpha $**:
   - If **$ p \leq \alpha $** → Reject $ H_0 $ (significant result).
   - If **$ p > \alpha $** → Fail to reject $ H_0 $ (not enough evidence).

---

## **🔹 3. Example: Quality Control for Cereal Box Filling**
**Scenario:**  
A cereal manufacturer tests whether its **filling machine** is discharging a **mean of 12 oz** per box. A random sample of **100 boxes** is collected.

**Given Data:**
- **$ n = 100 $** (Large sample)
- **$ \bar{x} = 11.851 $** (Sample mean)
- **$ s = 0.33 $** (Sample standard deviation)
- **$ \alpha = 0.01 $** (Significance level)
- **$ H_0: \mu = 12 $**
- **$ H_a: \mu \neq 12 $** (Two-tailed test)

**Step 1: Compute Test Statistic**
\begin{align}
z = \frac{11.851 - 12}{0.33 / \sqrt{100}}
\end{align}

\begin{align}
z = \frac{-0.149}{0.033} = -4.52
\end{align}

**Step 2: Compute p-value**  
For a two-tailed test,  
\begin{align}
p = 2 \times P(Z < -4.52)
\end{align}

Using Z-tables,  
\begin{align}
P(Z < -4.52) \approx 0.000003
\end{align}

\begin{align}
p = 2 \times 0.000003 = 0.000006
\end{align}

**Step 3: Compare with $ \alpha $**  
Since **$ p = 0.000006 < 0.01 $**, **reject $ H_0 $**.  
📌 **Conclusion:** The mean filling amount is significantly **less than** 12 oz. The machine needs adjustment.

## **🔹 4. Example: Hospital Length of Stay**
**Scenario:**  
A hospital administrator suspects the **average patient stay** has decreased due to a new **managed healthcare system**. Previously, the mean **Length of Stay (LOS) = 5 days**.

\begin{align}

p = P(Z < -1.52)
\end{align}
From Z-tables:
\begin{align}
P(Z < -1.52) = 0.064
\end{align}

**Step 3: Compare with $ \alpha $**  
Since **$ p = 0.064 > 0.05 $**, we **fail to reject $ H_0 $**.  
📌 **Conclusion:** There is **insufficient evidence** to conclude that the hospital stay has decreased.

## **🔹 5. Python Code for Hypothesis Testing**
Let's implement the cereal box example in Python.

```python
import scipy.stats as stats

# Given data
x_bar = 11.851  # Sample mean
mu_0 = 12       # Hypothesized mean
s = 0.33        # Sample standard deviation
n = 100         # Sample size
alpha = 0.01    # Significance level

# Compute test statistic
z = (x_bar - mu_0) / (s / (n ** 0.5))

# Compute p-value for two-tailed test
p_value = 2 * (1 - stats.norm.cdf(abs(z)))

# Output results
print(f"Test Statistic (z): {z:.2f}")
print(f"p-value: {p_value:.6f}")

# Decision
if p_value < alpha:
    print("Reject H0: The machine is underfilling the boxes.")
else:
    print("Fail to Reject H0: No strong evidence of underfilling.")
```

# **📌 7.5 Test of Hypothesis About a Population Mean: Student’s t-Statistic**


---

## **🔹 1. Overview**
When conducting hypothesis tests for a **population mean $ \mu $** using small samples ($ n < 30 $), we must use the **Student’s t-distribution** instead of the normal $ z $-distribution because:
1. **The Central Limit Theorem does not guarantee normality** for small samples.
2. **The population standard deviation $ \sigma $ is unknown**, so we use the sample standard deviation $ s $, which adds more variability to the test statistic.

📌 **Test Statistic for Small Samples (t-test):**
\begin{align}
t = \frac{\bar{x} - \mu_0}{s / \sqrt{n}}
\end{align}
- $ \bar{x} $ = Sample mean  
- $ \mu_0 $ = Hypothesized population mean under $ H_0 $  
- $ s $ = Sample standard deviation  
- $ n $ = Sample size  
- **$ t $-distribution** with $ df = n - 1 $ degrees of freedom.

---

## **🔹 2. Conditions for Using the t-Test**
✅ A **random sample** is selected from the target population.  
✅ The **population distribution is approximately normal** (this assumption is crucial for small samples).  

---

## **🔹 3. Steps for a Small-Sample t-Test**
1️⃣ **State the hypotheses**:  
   - **Null Hypothesis ($ H_0 $)**: The population mean equals a specified value.  
   - **Alternative Hypothesis ($ H_a $)**: The population mean **differs, is greater, or is less** than the specified value.

2️⃣ **Set the significance level ($ \alpha $)** (e.g., 0.05 or 0.01).

3️⃣ **Compute the test statistic**:
   \begin{align}
   t = \frac{\bar{x} - \mu_0}{s / \sqrt{n}}
   \end{align}

4️⃣ **Determine the rejection region**:
   - Use **t-tables** or statistical software to find the critical $ t $-value.
   - For a **two-tailed test**, reject $ H_0 $ if $ |t| > t_{\alpha/2, df} $.
   - For a **one-tailed test**, reject $ H_0 $ if $ t < -t_{\alpha, df} $ (left-tailed) or $ t > t_{\alpha, df} $ (right-tailed).

5️⃣ **Compute the p-value**:
   - Use t-tables or software to find the probability associated with the computed $ t $-value.

6️⃣ **Compare p-value to $ \alpha $**:
   - If **$ p \leq \alpha $** → Reject $ H_0 $ (significant result).
   - If **$ p > \alpha $** → Fail to reject $ H_0 $ (insufficient evidence).

---

## **🔹 4. Example: Transformer Parts Production**
**Scenario:**  
A machine produces an **average of 15.5 parts per hour**. After an overhaul, a **random sample of 17 one-hour periods** was taken to check if the average production changed.

**Given Data:**
- **$ n = 17 $** (Small sample)
- **$ \bar{x} = 14.8 $** (Sample mean)
- **$ s = 1.3 $** (Sample standard deviation)
- **$ \alpha = 0.05 $** (Significance level)
- **$ H_0: \mu = 15.5 $**
- **$ H_a: \mu \neq 15.5 $** (Two-tailed test)

**Step 1: Compute Test Statistic**
\begin{align}
t = \frac{14.8 - 15.5}{1.3 / \sqrt{17}}
\end{align}

\begin{align}
t = \frac{-0.7}{0.315} = -2.22
\end{align}

**Step 2: Find the Critical t-Value**  
Degrees of Freedom $ df = 17 - 1 = 16 $.  
From **t-tables**, for **$ \alpha = 0.05 $ (two-tailed test)**, **$ t_{0.025, 16} = 2.120 $**.

**Step 3: Compare with Rejection Region**  
Since **$ |t| = 2.22 > 2.120 $**, we **reject $ H_0 $**.

📌 **Conclusion:** The machine's production rate has significantly changed after the overhaul.

## **🔹 5. Example: Air Pollution Standards for a New Engine**

A car manufacturer wants to verify that a **new engine emits less than 20 ppm** of carbon. A sample of **10 engines** was tested.

- **$ s = 3.4 $**
- **$ \alpha = 0.05 $** (Significance level)
- **$ H_0: \mu = 20 $**
- **$ H_a: \mu < 20 $** (One-tailed test)

**Step 1: Compute Test Statistic**
\begin{align}
t = \frac{17.57 - 20}{3.4 / \sqrt{10}}
\end{align}

\begin{align}
t = \frac{-2.43}{1.075} = -2.26
\end{align}

**Step 2: Find the Critical t-Value**  
Degrees of Freedom $ df = 10 - 1 = 9 $.

## **🔹 6. Python Code for the Air Pollution Test**
Let's implement the t-test in Python.

```python
import scipy.stats as stats

# Given data
x_bar = 17.57  # Sample mean
mu_0 = 20      # Hypothesized mean
s = 3.4        # Sample standard deviation
n = 10         # Sample size
alpha = 0.05   # Significance level

# Compute test statistic
t = (x_bar - mu_0) / (s / (n ** 0.5))

# Compute p-value for one-tailed test
p_value = stats.t.cdf(t, df=n-1)

# Output results
print(f"Test Statistic (t): {t:.2f}")
print(f"p-value: {p_value:.4f}")

# Decision
if p_value < alpha:
    print("Reject H0: The new engine meets the pollution standard.")
else:
    print("Fail to Reject H0: Insufficient evidence to conclude compliance.")
```

# **📌 7.6 Large-Sample Test of Hypothesis About a Population Proportion**


---

## **🔹 1. Overview**
A **large-sample hypothesis test for a population proportion** is commonly used in **polls, surveys, and quality control studies**. It allows us to determine whether a sample proportion $ \hat{p} $ significantly differs from a hypothesized population proportion $ p_0 $.

**Key Use Cases:**
- Testing whether **at least 50% of customers prefer a product**.
- Evaluating whether **a defect rate is below a certain threshold**.
- Determining whether **a survey proportion differs from historical trends**.

---

## **🔹 2. Conditions for Using the Large-Sample Test**
✅ A **random sample** is selected from a **binomial population**.  
✅ The **sample size is large**:  
   - **Both $ np_0 $ and $ nq_0 $ must be at least 15**, where:
     \begin{align}
     q_0 = 1 - p_0
     \end{align}

This ensures that the normal approximation to the binomial distribution is valid.

---

## **🔹 3. Test Statistic for Large-Sample Proportion Tests**
For large samples, the sampling distribution of $ \hat{p} $ (sample proportion) follows a normal approximation:

\begin{align}
z = \frac{\hat{p} - p_0}{\sqrt{\frac{p_0 q_0}{n}}}
\end{align}

where:
- $ \hat{p} $ = Sample proportion $ = \frac{x}{n} $ (where $ x $ is the number of successes).
- $ p_0 $ = Hypothesized population proportion.
- $ q_0 = 1 - p_0 $.
- $ n $ = Sample size.

---

## **🔹 4. Steps for a Large-Sample Proportion Test**

1️⃣ **State the hypotheses**:  
   - **Null Hypothesis ($ H_0 $)**: The population proportion equals a specified value.
   - **Alternative Hypothesis ($ H_a $)**:
     - **Two-tailed test**: $ p \neq p_0 $
     - **One-tailed test (greater than)**: $ p > p_0 $
     - **One-tailed test (less than)**: $ p < p_0 $

2️⃣ **Select a significance level $ \alpha $** (e.g., 0.05 or 0.01).


3️⃣ **Compute the test statistic $ z $**:
   \begin{align}
   z = \frac{\hat{p} - p_0}{\sqrt{\frac{p_0 q_0}{n}}}
   \end{align}

4️⃣ **Find the rejection region**:
   - **For two-tailed tests**, reject $ H_0 $ if $ |z| > z_{\alpha/2} $.
   - **For one-tailed tests**, reject $ H_0 $ if:
     - **Right-tailed:** $ z > z_{\alpha} $.
     - **Left-tailed:** $ z < -z_{\alpha} $.

5️⃣ **Compute the p-value**:
   - Use normal probability tables or Python to find the probability associated with the computed $ z $-value.

6️⃣ **Make a decision**:
   - If $ p $-value **≤** $ \alpha $, **reject $ H_0 $** (significant result).
   - If $ p $-value **>** $ \alpha $, **fail to reject $ H_0 $** (insufficient evidence).

---

## **🔹 5. Example: Insider Trading and Stock Price Increases**
**Scenario:**  
An investor suspects that **insider trading affects stock prices**. Over a **one-year period**, they observe **576 insider purchases**, of which **327** resulted in the stock price increasing the next day.

**Given Data:**
- **$ n = 576 $** (sample size)
- **$ x = 327 $** (number of "successes"—stock price increases)
- **$ \hat{p} = \frac{327}{576} = 0.5677 $** (sample proportion)
- **$ p_0 = 0.5 $** (null hypothesized proportion)
- **$ \alpha = 0.05 $** (significance level)

### **Step 1: Compute Test Statistic**
\begin{align}
z = \frac{0.5677 - 0.5}{\sqrt{\frac{0.5 \times 0.5}{576}}}
\end{align}
\begin{align}
z = \frac{0.0677}{0.02041} = 3.32
\end{align}

### **Step 2: Find the Critical Value**
- **For a two-tailed test at $ \alpha = 0.05 $**, the critical values from the normal table are **$ \pm1.96 $**.

### **Step 3: Compare with the Rejection Region**
Since **$ z = 3.32 > 1.96 $**, we **reject $ H_0 $**.

📌 **Conclusion:** At the **5% significance level**, there is strong evidence that **insider trading significantly affects stock prices**.

## **🔹 6. Example: Defective Battery Rate**
A battery manufacturer claims that **less than 5% of its batteries are defective**. A **random sample of 300 batteries** finds **10 defective units**.

**Given Data:**
- **$ n = 300 $**
- **$ x = 10 $**
- **$ \hat{p} = \frac{10}{300} = 0.0333 $**
- **$ p_0 = 0.05 $**
- **$ \alpha = 0.01 $** (significance level)

### **Step 1: Compute Test Statistic**
\begin{align}
z = \frac{0.0333 - 0.05}{\sqrt{\frac{0.05 \times 0.95}{300}}}
\end{align}
\begin{align}
z = \frac{-0.0167}{0.0126} = -1.33
\end{align}

### **Step 2: Find the Critical Value**
- **For a left-tailed test at $ \alpha = 0.01 $**, the critical value from the normal table is **$ -2.33 $**.

### **Step 3: Compare with the Rejection Region**
Since **$ z = -1.33 > -2.33 $**, we **fail to reject $ H_0 $**.

📌 **Conclusion:** At the **1% significance level**, there is **insufficient evidence** to conclude that the defective rate is **below 5%**.

## **🔹 7. Python Code for Proportion Tests**
Let's implement the **insider trading hypothesis test** in Python.

```python
import scipy.stats as stats
import numpy as np

# Given data
n = 576  # Sample size
x = 327  # Number of successes
p_hat = x / n  # Sample proportion
p_0 = 0.5  # Null hypothesized proportion
alpha = 0.05  # Significance level

# Compute test statistic
z = (p_hat - p_0) / np.sqrt((p_0 * (1 - p_0)) / n)

# Compute p-value for a two-tailed test
p_value = 2 * (1 - stats.norm.cdf(abs(z)))

# Output results
print(f"Test Statistic (z): {z:.2f}")
print(f"p-value: {p_value:.4f}")

# Decision
if p_value < alpha:
    print("Reject H0: Insider trading affects stock prices.")
else:
    print("Fail to Reject H0: Insufficient evidence to conclude insider trading impact.")
```
