## **🔹 8.1 Identifying the Target Parameter**
Before diving into hypothesis tests and confidence intervals, we must first determine the **target parameter** for comparing two populations.

### **🔹 1. When to Use Two-Sample Inferences?**
When we compare:
- **Two population means** → e.g., comparing mean sales prices of city vs. suburban homes.
- **Two population proportions** → e.g., comparing voting preferences in two regions.
- **Two population variances** → e.g., comparing the variability in battery life between two brands.

| **Target Parameter** | **Key Words in Problem** | **Type of Data** |
|----------------|-----------------|----------------|
| Difference in means $ \mu_1 - \mu_2 $ | Mean difference, compare averages | Quantitative |
| Difference in proportions $ p_1 - p_2 $ | Proportions, percentages, rates | Qualitative |
| Ratio of variances $ \sigma_1^2 / \sigma_2^2 $ | Compare variability, spread | Quantitative |


---

## **🔹 8.2 Comparing Two Population Means: Independent Sampling**
This section focuses on comparing **two population means** using either:
1. **Large samples ($ n_1, n_2 \geq 30 $) → Normal (z) statistic**
2. **Small samples ($ n_1, n_2 < 30 $) → Student’s t-statistic**

### **1️⃣ Large-Sample Confidence Interval for $ \mu_1 - \mu_2 $**
**Scenario:**  
An economist investigates whether **Japanese cars are priced higher in Japan than in the U.S.** Independent samples of **50 car sales in both countries** were collected, and prices (converted to USD) were compared.

#### **Step 1: Define Hypotheses**
- **$ H_0: \mu_1 = \mu_2 $** (No difference in mean prices)
- **$ H_a: \mu_1 \neq \mu_2 $** (Mean prices differ)

#### **Step 2: Compute Confidence Interval**
\begin{align}
\left( \bar{x}_1 - \bar{x}_2 \right) \pm z_{\alpha/2} \cdot \sqrt{\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}}
\end{align}
Where:
- $ \bar{x}_1, \bar{x}_2 $ = Sample means
- $ s_1, s_2 $ = Sample standard deviations
- $ n_1, n_2 $ = Sample sizes
- $ z_{\alpha/2} $ = Critical value from z-table (e.g., 1.96 for 95% CI)

**Example Calculation:**
- $ \bar{x}_1 = 27.1 $, $ s_1 = 2.3 $ (U.S.)
- $ \bar{x}_2 = 26.8 $, $ s_2 = 2.5 $ (Japan)
- $ n_1 = 50 $, $ n_2 = 50 $

\begin{align}
(27.1 - 26.8) \pm 1.96 \cdot \sqrt{\frac{2.3^2}{50} + \frac{2.5^2}{50}}
\end{align}

\begin{align}
0.3 \pm 1.96 \cdot 0.455
\end{align}

\begin{align}
0.3 \pm 0.89
\end{align}

📌 **Confidence Interval**: **(-0.59, 1.19)** → Since 0 is in the interval, **no significant difference exists**.

---

### **2️⃣ Large-Sample Hypothesis Test for $ \mu_1 - \mu_2 $**
The test statistic follows:

\begin{align}
z = \frac{\left( \bar{x}_1 - \bar{x}_2 \right) - \text{Hypothesized Difference} }{\sqrt{\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}}}
\end{align}

Using the previous example:
\begin{align}
z = \frac{0.3 - 0}{0.455} = 0.66
\end{align}

- **Rejection region for $ \alpha = 0.05 $**: Reject $ H_0 $ if $ |z| > 1.96 $.
- **Since $ |0.66| < 1.96 $, fail to reject $ H_0 $.**  

📌 **Conclusion**: No strong evidence that Japanese cars are priced higher in Japan.


---

### **3️⃣ Small-Sample t-Test for $ \mu_1 - \mu_2 $**
If **$ n_1, n_2 < 30 $**, use **Student’s t-distribution**.

**Conditions:**
✅ Samples must be independent.  
✅ Populations should be **normally distributed**.  
✅ **Equal variances assumption** → Use **pooled variance estimator**:

\begin{align}
s_p^2 = \frac{(n_1 - 1)s_1^2 + (n_2 - 1)s_2^2}{n_1 + n_2 - 2}
\end{align}

**Test Statistic:**
\begin{align}
t = \frac{\bar{x}_1 - \bar{x}_2}{s_p \sqrt{\frac{1}{n_1} + \frac{1}{n_2}}}
\end{align}
with **$ df = n_1 + n_2 - 2 $**.

If **variances are unequal**, use **Welch’s t-test**, which adjusts degrees of freedom.

## **🔹 8.4 Comparing Two Population Proportions: Independent Sampling**

### **🔹 Introduction**
When comparing two population proportions, we are interested in testing whether there is a **significant difference** between them. For example, a manufacturer wants to compare the potential market for its products in the **Northeastern U.S.** and the **Southeastern U.S.**.  

We can estimate the difference in population proportions using a **large-sample confidence interval** and conduct a **hypothesis test** to determine if the difference is statistically significant.

When comparing two proportions $ p_1 $ and $ p_2 $, the test statistic follows:

\begin{align}
z = \frac{(\hat{p}_1 - \hat{p}_2)}{\sqrt{\hat{p}(1-\hat{p}) \left(\frac{1}{n_1} + \frac{1}{n_2}\right)}}
\end{align}

where:
- $ \hat{p}_1 = x_1/n_1 $, $ \hat{p}_2 = x_2/n_2 $ (sample proportions)
- $ \hat{p} = \frac{x_1 + x_2}{n_1 + n_2} $ (pooled estimate)

**Example: Election Polling**
- **45% (135 of 300 voters) prefer Candidate A in City 1**
- **40% (120 of 300 voters) prefer Candidate A in City 2**
- **Test if the difference is significant ($ \alpha = 0.05 $)**

\begin{align}
\hat{p} = \frac{135+120}{300+300} = 0.425
\end{align}
\
\begin{align}
z = \frac{0.45 - 0.40}{\sqrt{0.425 \times (1-0.425) \left(\frac{1}{300} + \frac{1}{300}\right)}}
\end{align}
\
\begin{align}
z = \frac{0.05}{0.036} = 1.39
\end{align}

Since **$ z = 1.39 < 1.96 $** (critical value at $ \alpha = 0.05 $), **fail to reject $ H_0 $**.  

📌 **Conclusion:** No significant difference in candidate preference.

---

### **🔹 Large-Sample Confidence Interval for $ p_1 - p_2 $**

#### **📌 Example: Personal Watercraft Market Study**
A manufacturer randomly selects **1,000 households in the Southeast (SE)** and **1,000 households in the Northeast (NE)** and asks whether they plan to buy a personal watercraft (PWC) in the next 5 years. The results are:

| Region | Sample Size ($ n $) | Households Buying PWC ($ x $) | Sample Proportion ($ \hat{p} $) |
|--------|------------------|--------------------|----------------------|
| **SE** | $ 1,000 $ | $ 110 $ | $ \hat{p}_1 = 110/1000 = 0.11 $ |
| **NE** | $ 1,000 $ | $ 80 $ | $ \hat{p}_2 = 80/1000 = 0.08 $ |

The point estimate of the difference in proportions is:

\begin{align}
\hat{p}_1 - \hat{p}_2 = 0.11 - 0.08 = 0.03
\end{align}

#### **📌 Step 1: Compute the Standard Error**
The **standard error** of $ \hat{p}_1 - \hat{p}_2 $ is:

\begin{align}
SE = \sqrt{\frac{\hat{p}_1 (1 - \hat{p}_1)}{n_1} + \frac{\hat{p}_2 (1 - \hat{p}_2)}{n_2}}
\end{align}

\begin{align}
SE = \sqrt{\frac{0.11(1 - 0.11)}{1000} + \frac{0.08(1 - 0.08)}{1000}}
\end{align}

\begin{align}
SE = \sqrt{\frac{0.11 \times 0.89}{1000} + \frac{0.08 \times 0.92}{1000}}
\end{align}

\begin{align}
SE = \sqrt{0.0000979 + 0.0000736} = \sqrt{0.0001715} = 0.0131
\end{align}

#### **📌 Step 2: Compute the Confidence Interval**
Using a **95% confidence level**, the critical value $ z_{\alpha/2} $ is **1.96**:

\begin{align}
(\hat{p}_1 - \hat{p}_2) \pm z_{\alpha/2} \cdot SE
\end{align}

\begin{align}
0.03 \pm 1.96 \times 0.0131
\end{align}

\begin{align}
0.03 \pm 0.0257
\end{align}

\begin{align}
(0.0043, 0.0557)
\end{align}

#### **📌 Interpretation**
We are **95% confident** that the true difference in proportions of households planning to buy a PWC in the **Southeast vs. Northeast** is **between 0.43% and 5.57%**. Since the confidence interval does **not** include **0**, we **have evidence** that the proportion of PWC buyers is higher in the Southeast.

---

### **🔹 Large-Sample Hypothesis Test for $ p_1 - p_2 $**
Now, we test:

\begin{align}
H_0: p_1 = p_2 \quad \text{(No difference)}
\end{align}
\begin{align}
H_a: p_1 \neq p_2 \quad \text{(Difference exists)}
\end{align}

#### **📌 Step 1: Compute the Pooled Proportion**
If $ H_0 $ is true, the best estimate for $ p $ is the **pooled proportion**:

\begin{align}
\hat{p} = \frac{x_1 + x_2}{n_1 + n_2}
\end{align}

\begin{align}
\hat{p} = \frac{110 + 80}{1000 + 1000} = \frac{190}{2000} = 0.095
\end{align}

#### **📌 Step 2: Compute the Test Statistic**
The test statistic is:

\begin{align}
z = \frac{(\hat{p}_1 - \hat{p}_2)}{\sqrt{\hat{p} (1-\hat{p}) \left( \frac{1}{n_1} + \frac{1}{n_2} \right)}}
\end{align}

\begin{align}
z = \frac{(0.11 - 0.08)}{\sqrt{0.095(1 - 0.095) \left( \frac{1}{1000} + \frac{1}{1000} \right)}}
\end{align}

\begin{align}
z = \frac{0.03}{\sqrt{0.095 \times 0.905 \times 0.002}}
\end{align}

\begin{align}
z = \frac{0.03}{\sqrt{0.000172}} = \frac{0.03}{0.0131} = 2.29
\end{align}

#### **📌 Step 3: Determine the p-Value**
Using a **z-table**, the probability associated with $ z = 2.29 $ is **0.0110**. Since this is a **two-tailed test**, we double this value:

\begin{align}
p\text{-value} = 2 \times 0.0110 = 0.022
\end{align}

#### **📌 Step 4: Conclusion**
Since **$ p = 0.022 < 0.05 $ (at $ \alpha = 0.05 $)**, we **reject $ H_0 $**.  
This provides **statistically significant** evidence that the proportion of **PWC buyers differs** between the Southeast and Northeast.

### **🔹 Example: Comparing Car Repair Rates**
A **consumer advocacy group** wants to compare the **proportions of two car models** needing **major repairs** (more than \$500) within 2 years.  

| Model | Sample Size ($ n $) | Owners with Repairs ($ x $) | Sample Proportion ($ \hat{p} $) |
|--------|------------------|--------------------|----------------------|
| **Model 1** | $ 400 $ | $ 53 $ | $ \hat{p}_1 = 53/400 = 0.1325 $ |
| **Model 2** | $ 500 $ | $ 78 $ | $ \hat{p}_2 = 78/500 = 0.156 $ |

\begin{align}
\hat{p}_1 - \hat{p}_2 = 0.1325 - 0.156 = -0.0235
\end{align}

#### **📌 Step 1: Compute the Pooled Proportion**
\begin{align}
\hat{p} = \frac{x_1 + x_2}{n_1 + n_2} = \frac{53 + 78}{400 + 500} = \frac{131}{900} = 0.1456
\end{align}

#### **📌 Step 2: Compute the Test Statistic**
\begin{align}
z = \frac{-0.0235}{\sqrt{0.1456(1 - 0.1456) \left( \frac{1}{400} + \frac{1}{500} \right)}}
\end{align}

\begin{align}
z = \frac{-0.0235}{\sqrt{0.1456 \times 0.8544 \times 0.0045}}
\end{align}

\begin{align}
z = \frac{-0.0235}{0.0157} = -1.50
\end{align}

#### **📌 Step 3: Compute the p-Value**
From the **z-table**, the probability associated with $ z = -1.50 $ is **0.0668**. Since this is **two-tailed**, the p-value is:

\begin{align}
p = 2 \times 0.0668 = 0.1336
\end{align}

#### **📌 Conclusion**
Since **$ p = 0.1336 > 0.05 $**, we **fail to reject $ H_0 $**.  
There is **insufficient evidence** that the repair rates differ.

### **🔹 Summary**
| **Test** | **Test Statistic** | **Condition** |
|------------|----------------|--------------|
| **Confidence Interval for $ p_1 - p_2 $** | $ (\hat{p}_1 - \hat{p}_2) \pm z_{\alpha/2} SE $ | Large samples $ np, nq \geq 15 $ |
| **Hypothesis Test for $ p_1 - p_2 $** | $ z = \frac{(\hat{p}_1 - \hat{p}_2)}{SE} $ | Large samples $ np, nq \geq 15 $ |

📌 **Use Fisher’s Exact Test for Small Samples.**

## **🔹 8.6 Comparing Two Population Variances**
When comparing two variances $ \sigma_1^2 $ and $ \sigma_2^2 $, we use the **F-statistic**:

\begin{align}
F = \frac{s_1^2}{s_2^2}
\end{align}

with degrees of freedom:
- **$ df_1 = n_1 - 1 $**
- **$ df_2 = n_2 - 1 $**

If **$ F > F_{\alpha, df_1, df_2} $** or **$ F < F_{1-\alpha, df_1, df_2} $**, reject $ H_0 $.

---

## **🔹 Summary Table: Comparing Two Populations**
| **Comparison Type** | **Test Statistic** | **Condition** |
|----------------|-------------------|----------------------|
| **Means ($ \mu_1 - \mu_2 $)** | $ z $ or $ t $ | Large or small sample |
| **Proportions ($ p_1 - p_2 $)** | $ z $ | Large sample ($ np \geq 15 $) |
| **Variances ($ \sigma_1^2 / \sigma_2^2 $)** | $ F $ | Independent samples |
